```sql
-- Count the number of customers for each country and order by customer count in descending order
SELECT `country`, COUNT(`customer_id`) AS `customer_count`
FROM `workspace`.`default`.customers_cvs
GROUP BY `country`
ORDER BY `customer_count` DESC;


-- Count the total number of products for each category in the Products table
SELECT 
    `category`, 
    COUNT(`product_id`) AS total_products 
FROM 
    `workspace`.`default`.`products_cvs` 
GROUP BY 
    `category`;
```
