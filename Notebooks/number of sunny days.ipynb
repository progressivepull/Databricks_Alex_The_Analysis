To determine the number of rainy days per city, we would look for records where precipitation is recorded and then count the number of distinct days with precipitation for each city. Here's the SQL query to get that information:

```sql
SELECT 
    `city_name`, 
    COUNT(DISTINCT `date`) AS `rainy_days`
FROM 
    `samples`.`accuweather`.`forecast_daily_calendar_metric`
WHERE 
    `has_precipitation` = false
GROUP BY 
    `city_name`
ORDER BY 
    `rainy_days` DESC;
```
